```python
from deepface import DeepFace
from pathlib import Path
```

```python
def get_png_files(directory_path):
    dir_path = Path(directory_path)
    png_files = list(dir_path.glob("*.png"))
    png_files = [str(file) for file in png_files]
    return png_files
```

```python
stem_folder : str = "../data/stable-diffusion/STEM"
mgmt_folder : str = "../data/stable-diffusion/MGMT"
```

```python
stem_files : list = get_png_files(stem_folder)
stem_files[0]
```

```python
for png_file in stem_files:
    try: 
        objs = DeepFace.analyze(
            img_path = png_file,
            actions = ['gender', 'race', 'emotion']
        )
        print(objs)
    except ValueError as e:
        print(e)
```
