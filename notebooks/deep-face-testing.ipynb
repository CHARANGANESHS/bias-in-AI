```python
ww_data = stats_WW
apac_data = stats_APAC
```

```python
def plot_heatmap(data, title):
    df = pd.DataFrame(data).T
    plt.figure(figsize=(8, 6))
    plt.title(title, fontsize=14)
    heatmap = plt.imshow(df, cmap="OrRd", aspect='auto')
    plt.colorbar(heatmap)
    plt.xticks(ticks=range(len(df.columns)), labels=df.columns, fontsize=12)
    plt.yticks(ticks=range(len(df.index)), labels=df.index, fontsize=12)
    plt.xlabel('Gender', fontsize=12)
    plt.ylabel('Race', fontsize=12)
    plt.grid(False)
    plt.show()
```

```python
plot_heatmap(ww_data, "WW Dataset - Heatmap (Race vs Gender)")
plot_heatmap(apac_data, "APAC Dataset - Heatmap (Race vs Gender)")
```
