```python
device = "cuda"
model_id: str = "runwayml/stable-diffusion-v1-5",
pipe = StableDiffusionPipeline.from_pretrained(
    model_id,
    torch_dtype=torch.float16 if device == "cuda" else torch.float32
).to(device)

def generate_image(
    prompt: str,
    output_path: str = "generated.png",
    height: int = 512,
    width: int = 512,
    num_inference_steps: int = 50,
    guidance_scale: float = 7.5,
):
       
    image = pipe(
        prompt,
        height=height,
        width=width,
        num_inference_steps=num_inference_steps,
        guidance_scale=guidance_scale
    ).images[0]

    image.save(output_path)
    print(f"Image saved to {output_path}")
```

```python
output_folder : str = "../data/hierarchy/junior"

for idx, desc in enumerate(junior_level):
    generate_image(
        prompt=desc,
        output_path=f"{output_folder}/{idx}.png",
    )
```

```python
output_folder : str = "../data/hierarchy/mid"

for idx, desc in enumerate(mid_level):
    generate_image(
        prompt=desc,
        output_path=f"{output_folder}/{idx + 33}.png",
    )
```

```python
output_folder : str = "../data/hierarchy/top"

for idx, desc in enumerate(top_level):
    generate_image(
        prompt=desc,
        output_path=f"{output_folder}/{idx + 66}.png",
    )
```
